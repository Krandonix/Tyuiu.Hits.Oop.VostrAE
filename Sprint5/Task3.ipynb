```
// КАРТА ЗАДАНИЙ СПРИНТА
// 1. Базовый и производные классы: BankAccount, SavingsAccount, CheckingAccount,
//    InvestmentAccount; дополнительная ступень — PremiumSavingsAccount.
// 2. Конструкторы и свойства: у каждого счёта есть конструктор и свойства get/set.
// 3. Наследование: простое — SavingsAccount : BankAccount;
//    многоуровневое — PremiumSavingsAccount : SavingsAccount : BankAccount.
//    Множественное наследование классов в C# запрещено; InvestmentAccount
//    наследует класс BankAccount и реализует несколько интерфейсов.
// 4. Полиморфизм: override в производных классах; перегрузка Deposit/Transfer.
// 5. Интерфейсы: IInvestable, IReportable, IInvestmentAccount.
// 6. Внедрение зависимости: InvestmentService получает интерфейс в конструкторе.
// 7. Коллекции и generic: List<T> и AccountRepository<T>.
// 8. Делегат и событие: BalanceChangedHandler и BalanceChanged.

// Делегат задаёт сигнатуру обработчика события изменения баланса.
public delegate void BalanceChangedHandler(BankAccount account, decimal oldBalance, decimal newBalance);

// Абстрактный базовый класс: общие данные и операции для всех видов счетов.
public abstract class BankAccount
{
    // Свойства с get/set; Balance разрешено изменять только внутри иерархии.
    public string AccountNumber { get; set; }
    public decimal Balance { get; protected set; }
    public string AccountType { get; set; }
 
    public string OwnerName { get; set; }
    public bool IsActive { get; set; }
    public DateTime CreatedDate { get; set; }

    // Событие уведомляет подписчиков после изменения баланса.
    public event BalanceChangedHandler BalanceChanged;
 
    // Конструктор записывает параметры через сеттеры свойств.
    public BankAccount(string accountNumber, decimal balance, string accountType)
    {
        AccountNumber = accountNumber;
        Balance = balance;
        AccountType = accountType;
 
        OwnerName = "Не указано";
        IsActive = true;
        CreatedDate = DateTime.Now;
    }

    // Общий метод изменения баланса: вызывает событие только при реальном изменении.
    protected void SetBalance(decimal newBalance)
    {
        decimal oldBalance = Balance;
        Balance = newBalance;
        if (oldBalance != newBalance)
            BalanceChanged?.Invoke(this, oldBalance, newBalance);
    }
 
    // virtual разрешает производным классам переопределить поведение.
    public virtual string GetInfo()
    {
        return $"Номер аккаунта: {AccountNumber}\n" + $"Тип аккаунта: {AccountType}\n" + $"Баланс: {Balance}\n" + $"Имя владельца: {OwnerName}\n" + $"Аккаунт активен: {IsActive}\n" + $"Дата создания: {CreatedDate}";
    }
 
    public virtual void Deposit(decimal amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("Сумма должна быть положительной");
            return;
        }
        SetBalance(Balance + amount);
    }
 
    // Перегрузка: то же имя Deposit, но другой набор параметров.
    public void Deposit(decimal amount, string description)
    {
        Console.WriteLine($"Операция: {description}");
        Deposit(amount);
    }
 
    public virtual bool Withdraw(decimal amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("Сумма должна быть положительной");
            return false;
        }
        if (amount > Balance)
        {
            Console.WriteLine("Недостаточно средств");
            return false;
        }
        SetBalance(Balance - amount);
        return true;
    }
 
    // Взаимодействие объектов: один счёт снимает деньги, второй получает их.
    public void Transfer(BankAccount account, decimal amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("Сумма перевода должна быть положительной");
            return;
        }
 
        if (!Withdraw(amount))
        {
            Console.WriteLine("Перевод не выполнен");
            return;
        }
 
        account.Deposit(amount);
        Console.WriteLine($"Перевод {amount} выполнен успешно");
    }
 
    // Ещё одна перегрузка — перевод с комментарием.
    public void Transfer(BankAccount account, decimal amount, string comment)
    {
        Console.WriteLine($"Комментарий к переводу: {comment}");
        Transfer(account, amount);
    }
 
    public void ActivateAccount()
    {
        IsActive = true;
        Console.WriteLine("Счёт активирован");
    }
 
    public void DeactivateAccount()
    {
        IsActive = false;
        Console.WriteLine("Счёт деактивирован");
    }
 
    public decimal GetBalance()
    {
        return Balance;
    }
}
 
// Простое наследование: один производный класс от одного базового.
public class SavingsAccount : BankAccount
{
    // Новые свойства сберегательного счёта.
    public decimal InterestRate { get; set; }
 
    public decimal MinimumBalance { get; set; }
    public bool AutoRenewal { get; set; }
    public int DepositCount { get; set; }
 
    public SavingsAccount(string accountNumber, decimal balance, decimal interestRate) : base(accountNumber, balance, "SavingsAccount")
    {
        InterestRate = interestRate;
 
        MinimumBalance = 1000;
        AutoRenewal = true;
        DepositCount = 0;
    }
 
    // override демонстрирует полиморфизм: добавляем сведения к базовому GetInfo().
    public override string GetInfo()
    {
        return base.GetInfo() + $"\nПроцентная ставка: {InterestRate}\nМинимальный баланс: {MinimumBalance}\nАвтоматическое продление: {AutoRenewal}\nКоличество депозитов: {DepositCount}";
    }
 
    // Переопределение: вклад зачисляется вместе с процентами.
    public override void Deposit(decimal amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("Сумма должна быть положительной");
            return;
        }
        decimal interest = amount * InterestRate / 100;
        SetBalance(Balance + amount + interest);
        DepositCount++;
 
        Console.WriteLine($"Внесено: {amount}, начислено процентов: {interest}");
    }
 
    public decimal CalculateInterest()
    {
        return Balance * InterestRate / 100;
    }
 
    public void ChangeInterestRate(decimal newRate)
    {
        if (newRate >= 0)
        {
            InterestRate = newRate;
        }
    }
 
    public bool CheckMinimumBalance()
    {
        return Balance >= MinimumBalance;
    }
}
 
// Многоуровневое наследование: BankAccount -> SavingsAccount -> PremiumSavingsAccount.
public class PremiumSavingsAccount : SavingsAccount
{
    public decimal CashbackPercent { get; set; }
    public string PremiumLevel { get; set; }
    public string PersonalManager { get; set; }
 
    public PremiumSavingsAccount(string accountNumber, decimal balance, decimal interestRate) : base(accountNumber, balance, interestRate)
    {
        AccountType = "PremiumSavingsAccount";
 
        CashbackPercent = 2;
        PremiumLevel = "Gold";
        PersonalManager = "Не назначен";
    }
 
    public override string GetInfo()
    {
        return base.GetInfo() + $"\nУровень: {PremiumLevel}" + $"\nКэшбэк: {CashbackPercent}" + $"\nПерсональный менеджер: {PersonalManager}";
    }
 
    public decimal CalculateCashback(decimal purchaseAmount)
    {
        return purchaseAmount * CashbackPercent / 100;
    }
 
    public void UpgradeLevel()
    {
        PremiumLevel = "Platinum";
        CashbackPercent = 3;
    }
 
    public void SetManager(string managerName)
    {
        PersonalManager = managerName;
    }
}
 
// Ещё один потомок BankAccount: отдельное поведение снятия денег.
public class CheckingAccount : BankAccount
{
    public decimal OverdraftLimit { get; set; }
 
    public decimal MonthlyFee { get; set; }
    public bool HasCard { get; set; }
    public string CardNumber { get; set; }
 
    public CheckingAccount(string accountNumber, decimal balance, decimal overdraftLimit) : base(accountNumber, balance, "CheckingAccount")
    {
        OverdraftLimit = overdraftLimit;
 
        MonthlyFee = 100;
        HasCard = true;
        CardNumber = "Не назначен";
    }
 
    public override string GetInfo()
    {
        return base.GetInfo() + $"\nЛимит овердрафта: {OverdraftLimit}" + $"\nЕжемесячная плата: {MonthlyFee}" + $"\nНаличие карты: {HasCard}" + $"\nНомер карты: {CardNumber}";
    }
 
    // Переопределение разрешает уйти в минус, но только в пределах овердрафта.
    public override bool Withdraw(decimal amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("Сумма должна быть положительной");
            return false;
        }
        if (Balance - amount < -OverdraftLimit)
        {
            Console.WriteLine("Отказано. Превышен лимит овердрафта");
            return false;
        }
        SetBalance(Balance - amount);
        Console.WriteLine($"Снято: {amount}");
        return true;
    }
 
    public void PayMonthlyFee()
    {
        if (Balance >= MonthlyFee)
        {
            SetBalance(Balance - MonthlyFee);
            Console.WriteLine($"Списана комиссия: {MonthlyFee}");
        }
    }
 
    public void SetCardNumber(string cardNumber)
    {
        CardNumber = cardNumber;
        HasCard = true;
    }
 
    public void ChangeMonthlyFee(decimal newFee)
    {
        if (newFee >= 0)
        {
            MonthlyFee = newFee;
        }
    }
}
 
// Контракт для операций инвестирования.
public interface IInvestable
{
    void Invest(decimal amount);
    decimal CalculateProfit();
}

// Отдельный контракт для формирования отчёта.
public interface IReportable
{
    string GetReport();
    void PrintReport();
}

// Интерфейс наследует два интерфейса. Это множественное наследование интерфейсов.
// Общий контракт для сервиса содержит только нужные ему возможности.
public interface IInvestmentAccount : IInvestable, IReportable
{
    string AccountNumber { get; }
    decimal Balance { get; }
}

// Класс наследует один базовый класс и реализует составной интерфейс.
public class InvestmentAccount : BankAccount, IInvestmentAccount
{
    public string AssetsList { get; set; }
 
    public decimal InvestmentProfit { get; set; }
    public string RiskLevel { get; set; }
    public bool AutoInvestment { get; set; }
 
    public InvestmentAccount(string accountNumber, decimal balance, string assetsList) : base(accountNumber, balance, "InvestmentAccount")
    {
        AssetsList = assetsList;
 
        InvestmentProfit = 0;
        RiskLevel = "Средний";
        AutoInvestment = false;
    }
 
    public override string GetInfo()
    {
        return base.GetInfo() + $"\nСписок активов: {AssetsList}" + $"\nДоходность: {InvestmentProfit}" + $"\nУровень риска: {RiskLevel}" + $"\nАвтоматическое инвестирование: {AutoInvestment}";
    }
 
    // Явная реализация: вызов доступен через IInvestable или IInvestmentAccount.
    void IInvestable.Invest(decimal amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("Сумма инвестиции должна быть положительной");
            return;
        }
 
        if (amount > Balance)
        {
            Console.WriteLine("Недостаточно средств для инвестирования");
            return;
        }
 
        SetBalance(Balance - amount);
        Console.WriteLine($"Инвестировано: {amount}");
    }
 
    // Явно реализованный метод не вызывается как investment.CalculateProfit().
    decimal IInvestable.CalculateProfit()
    {
        return InvestmentProfit;
    }
 
    // Отчёт тоже доступен через ссылку на интерфейс.
    string IReportable.GetReport()
    {
        return $"Инвестиционный отчёт\n" + $"Активы: {AssetsList}\n" + $"Прибыль: {InvestmentProfit}\n" + $"Риск: {RiskLevel}";
    }
 
    void IReportable.PrintReport()
    {
        Console.WriteLine(((IReportable)this).GetReport());
    }
 
    public void AddProfit(decimal profit)
    {
        if (profit > 0)
        {
            InvestmentProfit += profit;
            SetBalance(Balance + profit);
        }
    }
 
    public void ChangeRiskLevel(string newRiskLevel)
    {
        RiskLevel = newRiskLevel;
    }
}
 
// Внедрение зависимости через конструктор: сервис зависит от интерфейса,
// а не создаёт InvestmentAccount внутри себя.
public class InvestmentService
{
    private readonly IInvestmentAccount _account;

    public InvestmentService(IInvestmentAccount account)
    {
        _account = account ?? throw new ArgumentNullException(nameof(account));
    }

    public void InvestAndReport(decimal amount)
    {
        Console.WriteLine($"Счёт: {_account.AccountNumber}");
        _account.Invest(amount);
        _account.PrintReport();
    }
}

// Обобщённый (generic) репозиторий: T может быть BankAccount или его потомком.
public class AccountRepository<T> where T : BankAccount
{
    // Коллекция List<T> хранит счета; LINQ ниже ищет и суммирует их.
    private List<T> _accounts = new List<T>();
 
    public void Add(T account)
    {
        _accounts.Add(account);
    }
 
    public bool Remove(T account)
    {
        return _accounts.Remove(account);
    }
 
    public T FindByNumber(string accountNumber)
    {
        return _accounts.FirstOrDefault(a => a.AccountNumber == accountNumber);
    }
 
    public decimal GetTotalBalance()
    {
        return _accounts.Sum(a => a.Balance);
    }
 
    public void ShowAll()
    {
        foreach (var account in _accounts)
        {
            Console.WriteLine(account.GetInfo());
            Console.WriteLine("---");
        }
    }
}

public class Program
{
    public static void Main()
    {
        // ТЕМА 1: конструкторы, свойства get/set и методы производных классов.
        SavingsAccount savings = new SavingsAccount("S001", 10000, 5);
        savings.OwnerName = "Сергей";
        savings.Deposit(2000);
         
        Console.WriteLine("=== НАКОПИТЕЛЬНЫЙ СЧЁТ ===");
        Console.WriteLine(savings.GetInfo());
        Console.WriteLine($"Проценты: {savings.CalculateInterest()}");
        Console.WriteLine($"Минимальный баланс: {savings.CheckMinimumBalance()}");
         
        PremiumSavingsAccount premium = new PremiumSavingsAccount("P001", 50000, 7);
        premium.OwnerName = "Алексей";
        premium.SetManager("Иван Ященко");
        premium.UpgradeLevel();
         
        Console.WriteLine("\n=== ПРЕМИАЛЬНЫЙ СЧЁТ ===");
        Console.WriteLine(premium.GetInfo());
        Console.WriteLine($"Кэшбэк с покупки 10000: {premium.CalculateCashback(10000)}");
         
        CheckingAccount checking = new CheckingAccount("C001", 15000, 5000);
        checking.OwnerName = "Пётр";
        checking.SetCardNumber("1234-5678");
        checking.Withdraw(3000);
        checking.PayMonthlyFee();
         
        Console.WriteLine("\n=== ТЕКУЩИЙ СЧЁТ ===");
        Console.WriteLine(checking.GetInfo());
         
        InvestmentAccount investment = new InvestmentAccount("I001", 30000, "Акции, Облигации");
        investment.OwnerName = "Сергей";
        // ТЕМА 2: явная реализация. Метод Invest вызывается через интерфейс.
        IInvestmentAccount investmentContract = investment;
        // investment.Invest(20000); // Ошибка компиляции при явной реализации.
        investmentContract.Invest(20000);
        investment.AddProfit(5000);
        investment.ChangeRiskLevel("Высокий");
         
        Console.WriteLine("\n=== ИНВЕСТИЦИОННЫЙ СЧЁТ ===");
        Console.WriteLine(investment.GetInfo());
         
        Console.WriteLine("\n=== ОТЧЁТ ===");
        investmentContract.PrintReport();
         
        // Один объект доступен через разные интерфейсы.
        IInvestable investable = investment;
        Console.WriteLine($"\nПрибыль через интерфейс: {investable.CalculateProfit()}");
         
        IReportable reportable = investment;
        Console.WriteLine("\n=== ОТЧЁТ ЧЕРЕЗ ИНТЕРФЕЙС ===");
        reportable.PrintReport();

        // ТЕМА 3: внедрение зависимости через конструктор сервиса.
        Console.WriteLine("\n=== ВНЕДРЕНИЕ ЗАВИСИМОСТИ ===");
        InvestmentService investmentService = new InvestmentService(investmentContract);
        investmentService.InvestAndReport(1000);
         
        // ТЕМА 4: взаимодействие двух объектов BankAccount.
        Console.WriteLine("\n=== ПЕРЕВОД ===");
        savings.Transfer(checking, 1000);
        Console.WriteLine($"Баланс накопительного счёта: {savings.GetBalance()}");
        Console.WriteLine($"Баланс текущего счёта: {checking.GetBalance()}");
         
        // ТЕМА 5: полиморфизм. Вызов virtual/override зависит от реального типа.
        Console.WriteLine("\n=== ПОЛИМОРФИЗМ (ПЕРЕОПРЕДЕЛЕНИЕ) ===");
        // Одновременно это пример коллекции разных потомков одного класса.
        List<BankAccount> accounts = new List<BankAccount> { savings, premium, checking, investment };
         
        foreach (BankAccount acc in accounts)
        {
            acc.Deposit(1000);
            acc.Withdraw(500);
        }
         
        // ТЕМА 6: перегрузка — одно имя метода и разные параметры.
        Console.WriteLine("\n=== ПЕРЕГРУЗКА ===");
        savings.Deposit(500);
        savings.Deposit(500, "Пополнение с зарплаты");
        savings.Transfer(checking, 300);
        savings.Transfer(checking, 300, "Оплата подписки");
         
        // ТЕМА 7: generic-класс и коллекция List<T> внутри репозитория.
        Console.WriteLine("\n=== GENERIC-КЛАСС И КОЛЛЕКЦИЯ ===");
        AccountRepository<SavingsAccount> savingsRepo = new AccountRepository<SavingsAccount>();
        savingsRepo.Add(savings);
        savingsRepo.Add(premium);
        Console.WriteLine($"Общий баланс накопительных: {savingsRepo.GetTotalBalance()}");
         
        AccountRepository<BankAccount> allRepo = new AccountRepository<BankAccount>();
        allRepo.Add(savings);
        allRepo.Add(checking);
        allRepo.Add(investment);
        Console.WriteLine($"Общий баланс всех: {allRepo.GetTotalBalance()}");
         
        var found = allRepo.FindByNumber("C001");
        Console.WriteLine($"Найден счёт: {found?.AccountNumber}");

        // ТЕМА 8: делегат хранит метод-обработчик; += подписывает его на событие.
        Console.WriteLine("\n=== ДЕЛЕГАТ И СОБЫТИЕ ===");
        BalanceChangedHandler balanceLogger = (account, oldBalance, newBalance) =>
            Console.WriteLine($"Счёт {account.AccountNumber}: {oldBalance} -> {newBalance}");
        savings.BalanceChanged += balanceLogger;
        checking.BalanceChanged += balanceLogger;
        savings.Deposit(100);
        checking.Withdraw(50);
        // -= снимает подписку, если уведомления больше не нужны.
        savings.BalanceChanged -= balanceLogger;
        checking.BalanceChanged -= balanceLogger;
    }
}
```
